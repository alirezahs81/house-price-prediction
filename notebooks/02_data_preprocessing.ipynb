# 02 — Data Preprocessing

Explore the shared feature-engineering and sklearn preprocessing implementation. All learned imputation and encoding parameters are fit inside the pipeline, preventing preprocessing leakage across cross-validation folds.

```python
from src.data_loader import load_train, load_test
from src.preprocessing import prepare_features, transform_absence_categories, build_preprocessor
train = load_train()
X = transform_absence_categories(prepare_features(train))
print("Feature matrix before transformation:", X.shape)
display(X.head())
```

```python
preprocessor = build_preprocessor(X)
X_transformed = preprocessor.fit_transform(X)
print("Transformed matrix:", X_transformed.shape)
print("All finite:", __import__("numpy").isfinite(X_transformed).all())
```

## Design notes

- Numeric features use median imputation and standard scaling.
- Categorical features use most-frequent imputation and one-hot encoding.
- Unseen inference categories are ignored safely by the encoder.
- Feature engineering is deterministic and target-independent.
- The `Id` and `SalePrice` columns are excluded from predictors.
