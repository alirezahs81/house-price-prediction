# 02 — Data Preprocessing

Explore the shared feature-engineering and sklearn preprocessing implementation. The split is made before fitting imputers, scalers, or encoders so validation data cannot influence learned preprocessing statistics.

```python
from sklearn.model_selection import train_test_split
from src.data_loader import load_train
from src.preprocessing import prepare_features, transform_absence_categories, build_preprocessor

train = load_train()
y = train["SalePrice"]
X = transform_absence_categories(prepare_features(train))
X_train, X_validation = train_test_split(
    X, test_size=0.2, random_state=42
)
print("Training features:", X_train.shape)
print("Validation features:", X_validation.shape)
print("Target leaked into features:", "SalePrice" in X.columns)
```

```python
preprocessor = build_preprocessor(X_train)
X_train_transformed = preprocessor.fit_transform(X_train)
X_validation_transformed = preprocessor.transform(X_validation)
import numpy as np
print("Training transformed shape:", X_train_transformed.shape)
print("Validation transformed shape:", X_validation_transformed.shape)
print("All validation values finite:", np.isfinite(X_validation_transformed).all())
```

## Design notes

- Numeric features use median imputation and standard scaling.
- Categorical features use most-frequent imputation and one-hot encoding.
- Unseen validation/inference categories are ignored safely by the encoder.
- The preprocessor is fitted only on training rows; validation rows are transform-only.
- Feature engineering is deterministic and target-independent.
- The Id and SalePrice columns are excluded from predictors.
